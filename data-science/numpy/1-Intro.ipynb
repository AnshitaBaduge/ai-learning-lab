## NumPy

NumPy(Numerical Python) is a multi-dimensional array library. What that means is you can use numpy to store all sorts of data, 1D, 4D, etc. Its the core library for numerical computing in Python. Almost every data science library (Pandas, Scikit-learn, TensorFlow) is built on top of it.

-> Why not use python lists? how are lists different from numpy?
   The answer is speed. Python lists are slow for math (they store objects, not raw numbers) meanwhile,
   NumPy arrays store data in contiguous memory blocks of a fixed type → much faster

   ![alt text](image.png)

   NumPy is faster to read less bytes of memory. 
   when we are iterating through each item in a numpy array, we don't have to do type checking each time. NumPy supports vectorized operations (no explicit loops needed).

   ![alt text](image-1.png)

```python
# load in NumPy (remember to pip install numpy first)

import numpy as np
```

The Basics

```python
a = np.array([1,3,5], dtype= 'int16')     # 1D array
print(a)

b = np.array([[2,4,6], [8,10,12]])
print(b)
```

```python
b.shape     # dimensions
```

```python
b.ndim     # number of dimensions
```

```python
b.size      # 6 -> total elements
```

```python
b.dtype     # data type, e.g. int64

# if we want to specify what type we want it to store, we can do for ex; a = np.array([1,2,3], dtype= 'int16')
```

```python
b.itemsize  # bytes per element
```

```python
b.nbytes   # total size
```
