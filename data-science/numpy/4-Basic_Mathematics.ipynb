## Broadcasting:
 lets NumPy perform operations on arrays of different shapes without manually looping, by "stretching" the smaller array.

```python
import numpy as np
a = np.array([1,2,3,4])
print(a)
```

```python
a + 2    # can also do a += 2
```

Rule: compare shapes from the right; dimensions must be equal, or one of them must be 1.

(2,3) + (3,)   -> works (treated as (1,3))

(2,3) + (2,1)  -> works

(2,3) + (2,)   -> fails

```python
a - 2
```

```python
a * 2
```

```python
a / 2
```

```python
b = np.array([1,0,1,0])
a + b
```

```python
a ** 2
```

```python
# take the sin
np.sin(a)   # also cos and others
```

## Vectorization:
replacing explicit Python loops with array-level operations. This is the single biggest reason NumPy is fast.

```python
# Slow way
result = []
for x in range(1000000):
    result.append(x**2)

# Vectorized way
x = np.arange(1000000)
result = x**2         # much faster, runs in C internally
```

Common vectorized operations:
-> np.sqrt(a)

-> np.exp(a)

-> np.log(a)

-> np.abs(a)

-> a + b, a - b, a * b, a / b     # all element-wise

Universal functions (ufuncs) are the vectorized functions NumPy provides (np.add, np.multiply, etc.) — they're what make broadcasting + vectorization fast under the hood.

# Linear Algebra

```python
a = np.ones((2,3))
print(a)

b = np.full((3,2), 2)
print(b)

np.matmul(a,b)     # matric multiplication
```

```python
# find the determinant
c = np.eye(3)
np.linalg.det(c)
```

'*' is element-wise, '@' is true matrix multiplication.

we can also do for eigenvalues, inverse of a matrix, trace, singular vector decomposition, matrix norm, etc...

A + B              # element-wise addition

A * B              # element-wise multiplication (NOT matrix mult!)

A @ B              # matrix multiplication (or np.matmul(A,B) / np.dot(A,B))

A.T                # transpose

np.linalg.inv(A)   # inverse

np.linalg.det(A)   # determinant

np.linalg.eig(A)   # eigenvalues/eigenvectors

np.linalg.solve(A, b)  # solve Ax = b

# Axis

Rule of thumb: axis=0 collapses rows (goes down), axis=1 collapses columns (goes across).

```python
m = np.array([[1,2,3],[4,5,6]])
m.sum(axis=0)   # sum down columns -> [5,7,9]
m.sum(axis=1)   # sum across rows  -> [6,15]
```
