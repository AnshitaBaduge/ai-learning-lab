## Git

Version Control System: to track changes to file over time, so that we can manage history and restore older states when required.

->When we write some code, do something and it's not the final file. We tends to make changes, file, file1, file2.. file_final...and its very difficult to know, WHAT changed, WHEN changed, WHO changed and WHY changed. 
Which is not only difficult but also not feasible and error prone too.

Version Control Types:
1] Centralized: Repository lives on server and server maintains its full history. Developer machine is temporary editable copy, which works only on latest changes.
