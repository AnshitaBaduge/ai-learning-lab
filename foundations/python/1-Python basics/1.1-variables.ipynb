Variables: elements in prgming used to store data that can be referenced and manipulated in a prgm.

In python, var r created when assign value & they don not need explicit declaration to reserve memory space.

```python
## Declaring and assigning var
age = 20
height = 5.4
name = "Anshita"
fav_color = input("What is your favorite color?")
is_stud = True

## Printing
print("age:",age)
print("height:",height)
print("Name:",name)

print(name + ' likes ' + fav_color)
```

```python
## type conversion
age_str= str(age)
print(age_str)
print(type(age_str))

print(type(name), "\n" )
# int(name), str cannot be just converted into int
```

```python
## Dynamic Typing
## Python allows the type of a variable to change as the prgm executes
var = 10 #int
print(var,type(var))

var = "Hello"
print(var,type(var))

var = 3.14
print(var,type(var))
```

```python
## Calculator

num1 = int(input("Enter a number:"))
num2 = int(input("Enter another number:"))

sum = num1 + num2
diff = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum is:",sum)
print("Difference is:",diff)
print("Product is:",product)
print("Quotient is:",quotient)
```
