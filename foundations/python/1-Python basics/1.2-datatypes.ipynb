```python
## string methods
course = 'Python for Beginners'
print(len(course))  # useful when to receive input from the user, within a limit. It's a general purpose function.

# to access the functions we use the dot operator. we refer to these functions as methods. 
# When a function belongs to something else or in specific to some kind of object, we refer to that function as a method.
"""
print("Alex") / len(Alex) → function, works independently
"alex".upper() → method, works attached to a specific value/object, and usually acts on that object's own data

Rule of thumb: if it's called with a dot (object.thing()), it's a method. If it's called on its own (thing()), it's a function.
"""

print(course.upper()) # this method does not modify or change our original string, in fact it create a new string and return it
print(course)

print(course.find('o')) # find method is case sensitive
print(course.find('O'))
print(course.find('Beginners'))

print(course.replace('Beginners', 'absolute beginners')) #replace is also case sensitive

print('Python' in course) 
# differnce between find method and in operator, is that find method returns the index of that character
# but in operator produces a boolean value.
```
