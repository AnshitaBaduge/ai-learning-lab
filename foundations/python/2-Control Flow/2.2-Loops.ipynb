```python
## while loopp
# The while loop continues to execute as long as the condition is true.

count = 0
while count<5:
    print(count)
    count += 1

print(" ")
i = 1
while i <= 5:
    print('*' * i)
    i += 1
```

```python
## loop control statements
#break: the break statement exists the loop permaturely.

for i in range(10):
    if (i==5):
        break
    print(i)
        
```

```python
## Guessing Game
secret_num = 9
i  = 0
guess_limit = 3
while i < guess_limit:
    guess = int(input("Guess: "))
    i += 1
    if guess == secret_num:
        print("You Win!!")
        break #when python interpreter see this, it going to immediately terminate the loop.
else:  # similar like if statements, our while statements can also have an else part.
    print("Sorry, You Failed")
```

```python
## Continue
# the continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass
# the pass statement is a null operation; it does nothing.

for i in range(5):
    if i==4:
        pass   #empty function, we can just write pass
    print(i)
```

```python
## Nested loops
# a loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}") # or also like print(f'({x}, {y})')
```

```python
numbers = [5,2,5,2,2]
# for num in numbers:
#     print('*' * num)
    
# by using inner loop
for num_i in numbers:
    res = ''
    for num in range(num_i):
        res += '*'
    print(res)    
```

```python
## Example: Calculate the sum of first N natural numbers using a while and for loop

#While loop
n = 10
sum = 0
count = 0
while count<=n:
    sum+=count
    count+=1
print("Sum of First",n,"numbers:",sum)    
```

```python
#For loop
n = 10
sum = 0
for i in range(n+1): # or for i in range(11)
    sum+=i
print("Sum of First",n,"numbers:",sum)    
```

```python
## Eg; Prime numbers between 1 and 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num,"is a prime number.")
            
        
```
