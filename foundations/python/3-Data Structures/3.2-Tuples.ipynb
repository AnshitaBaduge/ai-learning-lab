Tupels are ordered collections of item that are immutable. They are similar to lists but their immutability makes them different.

```python
## Tupel
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))

lst = list()
print(type(lst))
```

```python
nums = tuple([1,2,3,4,5]) # A list can be converted to a tuple.
nums
```

```python
lst = list((1,2,3.1,"a")) # A tupel can be converted to a list.
lst
```

```python
## Accessing tuple elements
print(nums)
nums[0]
```

```python
print(nums[-1])
print(nums[0:4])
print(nums[0::])
print(nums[::-1])
```

```python
# Tuple Operations
mixed_tuple = (1,2,3,4,"a","b",2.3)
concatenation_tuple = nums + mixed_tuple
print(concatenation_tuple)
```

```python
print(mixed_tuple*3)
print(nums*3)
```

```python
### Immutable nature of tuples i.e.,their elements cannot be changed once assigned.
lst = [1,2,3,4,5] #list
print(lst)
lst[1]="Anshita"  #list are mutable
print(lst)

#mixed_tuple[1] = "Anshita" ; Tuples are immutable
#print(mixed_tuple)
```

```python
## Tuple methods
print(nums.count(1))
print(nums.index(4)) #here index() return first index of value.
```

```python
## Packing and Unpacking tuple
packed_tuple = 1,"Hello", 3.14    #Packing
print(packed_tuple)
```

```python
## jaise abhi samjho coordinates = (1,2,3) hai, abhi hume inhe kisi calculation mein baar baar use karna hai, toh baar baar likhne se aacha,
#  hum assign karde, jaise x = coordinates[0] y = coordinates[1] z = coordinates[2] ye bahot lamba process ho raha hai,
a,b,c = packed_tuple     #Unpacking (hum python mein simply x,y,z = coordinates likh sakte hai)
print(a) 
print(b)
print(c)  #unpacking is not only limited to tuples, we can use this feature for lists as well.
```

```python
## Unpacking with *
nums=(1,2,3,4,5,6)
first,*middle,last = nums
print(first)
print(middle)
print(last)
```

```python
lst=[[1,2,3,4],[6,7,8,9],[3.14,"A"]]        #Nested List
print(lst[0][2])
print(lst[0][0:3])

lst=[[1,2,3,4],[6,7,8,9],(1,"Heyy",3.14,"A")]        # inside list there can be a tuple
print(lst[2][0:3])
```

```python
### Nested tuple
nested_tuple=((1,2,3),("a","b","c"),(True,False))    #Tuple inside a tuple
#Access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
