Sets are a built-ib data type to store collections of unique items. They are unordered and do not allow duplicate elements. Sets operations like union, intersection, difference and symmetric difference.

```python
## Create a set
empty_set = set()
print(empty_set)
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))

my_set=set([1,2,3,4,5,6,6,6,6,6,7])  #No duplicates
print(my_set)
```

```python
## Basic Sets Opeeration

#Adding and removing elements
my_set.add(8)
print(my_set)

my_set.remove(3)
print(my_set)
```

```python
my_set.discard(10)   #agar remove(10) likha hota toh error deta esliye discard use kra
my_set
```

```python
## pop method
removed_ele = my_set.pop()  # pop remove 1st element
print(removed_ele)
print(my_set)
```

```python
## clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set membership test
my_set = {1,2,3,4,5,6,7}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical operation
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Union
union_set = set1.union(set2)
print(union_set)

## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Difference
print(set1.difference(set2))

set1.difference_update(set2)
print(set1)   # set1 ki values difference wli ho gui
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

set2.difference(set1)
```

```python
## Symmetric difference
set1.symmetric_difference(set2)
```

```python
## Set Methods
set1 = {1,2,3,4,5}
set2 = {3,4,5}
print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
lst = [1,2,2,3,4,3,5,6]
# We can remove duplicate by simply converting list into set
set(lst)
```

```python
## Counting unique words in text
texts = "In this tut, we are discussing about sets"
words = texts.split()

# convert list of words into set to get unique words
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
