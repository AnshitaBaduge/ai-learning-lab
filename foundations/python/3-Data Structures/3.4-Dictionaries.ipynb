Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g. strings,numbers or tuples), while value can be of any type.

## Shallow Copy: A copy of the outer object only; nested/inner objects inside it are still shared references, not copied.

The core issue: list2 = list1 doesn't copy anything — both names point to the same list in memory. Change one, the other changes too.
ex: 
list1 = [1, 2, 3]
list2 = list1
list2.append(4)
print(list1)  # [1, 2, 3, 4] — changed too!

# Shallow copy (list1.copy()):
 makes a new outer list, but if it contains nested lists, those inner lists are still shared.
ex:
list1 = [[1, 2], [3, 4]]
list2 = list1.copy()
list2[0].append(99)
print(list1)   # [[1, 2, 99], [3, 4]] — inner list still shared!


## Deep Copy : A complete, independent copy — including all nested objects inside, recursively, all the way down.

# Deep copy (copy.deepcopy(list1)): 
 copies everything, including nested lists. Fully independent, no sharing at all.
ex:
list2 = copy.deepcopy(list1)
list2[0].append(99)
print(list1)   # unaffected

Rule: simple flat list → shallow copy is enough. Nested lists/dicts → use deepcopy.

```python
## Nested Dictionaries
students = {
    "stud1":{"name":"A","age":32},
    "stud2":{"name":"B","age":35}
}
print(students)
```

```python
## Access nested dictionaries elements
print(students["stud1"]["name"])
print(students["stud2"]["name"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for stud_id,stud_info in students.items():
    print(f"{stud_id}:{stud_info}")
    for key,value in stud_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprenhension
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition dictionary comprehension 
evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Use dictionary to count the frequency of elements in list

nums = [1,2,2,3,3,3,4,4,4,4]
freq = {}

for num in nums:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
print(freq)            
```

```python
## Merge 2 dictionaries into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```

```python
## example:
number = input("Phone: ")
digits_mapping = {
    "1" : "one",
    "2" : "two",
    "3" : "three",
    "4" : "four"
}
output = ""
for ch in number:
    output += digits_mapping.get(ch, "not valid") + " "
print(output)    
```

```python
## emoji converter
message = input(">")
words = message.split(' ')
emojis = {
    ":)" : "🙂",
    ":(" : "😔"
}
output = " "
for word in words:
    output += emojis.get(word, word) + " "
print(words)
print(output)
```
