```python
## Even or Odd
def even_or_odd(num):
    """This function finds if num is even or odd"""
    if num%2==0:
        print(f" Num {num} is even")
    else:
        print(f"Num {num} is odd.")   
        
## Call this function
even_or_odd(7)         #7 is an argument that we pass to the num parameter.
#when an function have parameters, we're obligated to pass a value for that parameter.

## parameters; are the holes or placeholders that we define in our function for receiving info
## and arguments; are the actual pieces of info that we supply to these functions.
```

```python
## function with multiple parameters
def add(a,b):
    "..."
    return a+b
res=add(9,5) ## 9 and 5 are positional arguments.(which means thier position or order matters)
print(res)
```

```python
### Default Parameters

# Normal para
def greet(name):
    print(f"Hello {name}, Welcome To The Paradise.")
    
greet("Anshita")    

# default para
def greet(name="Guest"):
    print(f"Hello {name}, Welcome To The Paradise.")
    
greet()  #no parameters
```

```python
### Variable Length Arguments
## Positional And Keywords Arguments

def print_nums(*Anshita):
    "..."
    for num in Anshita:
        print(num)
        
print_nums(3,1,2,"A",5,7,4.4,"B")    # the above star refers to all the positional arguments here.   
```

```python
## Positional Arguments
def print_nums(*args):
    for num in args:
        print(num)
        
print_nums(8,7,6,5,4,3,2,1.4)        
```

```python
## Keywords arguments; with these the position doesn't matter

def print_info(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
        
print_info(name="Anshita",age=20,country="India")    

## another example, cal_cost(total=50, shipping=5, discount=0.1) ; we specify what the parameter value represent  
# but for the most part use positional args, but if you're dealing with the functions that take multiple numerical values then see can we use kwagrs
# and also remember, if you're mixing args and kwagrs, you should always use pos agrs first and then kwargs.
```

```python
def print_info(*args,**kwargs):
    for num in args:
        print(f"Positional arguments: {num}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
        
print_info(1,2,3,name="Anshita",age=20,country="India")      # first give pos args and then keyword args
```
