Lambda functions are small anonymous functions defined using the lambda keyword. They can have any number of agruments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

```python
#Syntax
lambda arguments: expression
```

```python
# Normal Fun 
def add(a,b):
    return a+b

add(2,3)
```

```python
# Now Lambda Fun
add = lambda a,b:a+b
type(add)
print(add(2,3))
```

```python
even = lambda num:num%2==0
even(4)
```

```python
## Map
def square(num):
    return num**2
square(3)

## map- applies a function to all items in a list
nums=[1,2,3,4,5,6]
list(map(lambda x:x**2,nums))
```
