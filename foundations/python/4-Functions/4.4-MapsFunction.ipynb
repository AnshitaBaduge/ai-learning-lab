The map() function applies a given func to all items in an input list(or any other iterable) and returns a map object(an iterator). This is particulary useful for transforming data in a list comprehensively. 

```python
nums=[1,2,3,4,5,6]
def square(x):
    return x*x
list(map(square,nums))
```

```python
nums=[1,2,3,4,5,6]
list(map(lambda x:x**2,nums))
```

```python
### map multiple iterables
num1=[1,2,3,4,5]
num2=[6,7,8,9,10]
added_num=list(map(lambda x,y:x+y,num1,num2))
print(added_num)
```

```python
## map() to convert a list of strings to integers
# use map to convert strings to integers
str_nums = ['1','2','3','4','5']
int_nums = list(map(int, str_nums))
print(int_nums)
```

```python
str_words = ['a','b','c','d']
upper_words = list(map(str.upper,str_words))
print(upper_words)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name':'A','age':19},
    {'name':'B','age':20},
    {'name':'C','age':21},
    {'name':'D','age':22}
]
list(map(get_name,people))
```
