#data (variables) and behavior (functions) have been separate. OOP bundles them into one unit called an object, modeled from a blueprint called a class.
# ## Classes; 
-> capitalized the first letter( pascal naming convention) and no use of underscore for naming our classes.
-> for variables and functions we always use lower case letters and separate multiple words by using underscore.

```python
"""This module demonstrates the Point class."""
class Point:
    "..."
    def move(self):
        "..."
        print("Move.")
        
    def draw(self):
        "..."
        print("Draw.")
        
        
point1 = Point() # store object in a variable
point1.x = 10    # this objecvt can also have attributes and these attributes are variable that belong to a particular object.
point1.y = 20
print(point1.x)
point1.draw()

## Each object is a different instance of our point class
```

```python
class Student:
    "..."
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        "..."
        print(f"Name: {self.name}, Marks: {self.marks}")
```

#Above- 
# class Student: —> declares a new blueprint called Student
# def __init__(self, name, marks): —> the constructor (covered fully in 5.2)
# self.name = name —> stores name as an attribute on this specific object
# def display(self): —> a method: a function that belongs to the class and can access/use    the object's own data

```python
s1 = Student("Alex", 85) #s1 and s2 are two separate objects(also called instances)
s2 = Student("Priya", 92)

s1.display()   # Name: Alex, Marks: 85
s2.display()   # Name: Priya, Marks: 92; changing s1.marks never touches s2.marks
```

-> Class: The blueprint/template
-> Object / Instance: An actual thing built from the blueprint
-> Attribute: A variable that belongs to an object (self.name)
-> Method: A function that belongs to a class (display())
