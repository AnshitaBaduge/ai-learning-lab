## A constructor(__init__): is a function that gets called at the time of creating an object.

## __init__ is short for initialized; and this is a special method that Python automatically calls the moment you create a new object, you never call it yourself directly.

## self is a refernce to the current object

```python
class Student:
    "..."
    def __init__(self, name, marks):
        print("A new student object is being created!")
        self.name = name
        self.marks = marks
        

s1 = Student("Anshita", 87) # O/p - A new student object is being created!
```

# you never wrote s1.__init__(...). Just calling Student("Alex", 85) triggers it automatically. 
-> Its whole job is initializing (setting up) the object's starting data — hence the name "constructor" (it constructs the object's initial state).
-> refer to point.x and point.y example from previous module

# def __init__(self, name, marks):
# self refers to "the specific object currently being built/used." 
-> Without it, Python wouldn't know which object's name and marks you mean, since a class can produce many objects.
-> Python passes self automatically — you never type it in when calling; you only include it when defining methods.
-> s1 = Student("Alex", 85)   # Python internally treats this as Student.__init__(s1, "Alex", 85)

```python
## Default values in constructors:
class Stud:
    "..."
    def __init__(self, name, marks=0):
        self.name = name
        self.marks = marks

s1 = Stud("Alex")          # marks defaults to 0
s2 = Stud("Priya", 92)     # marks explicitly given
```

```python
## Constructor without parameters (still valid):
class Counter:
    "..."
    def __init__(self):
        self.count = 0   # Every object still gets initialized, just with fixed starting values instead of user-supplied ones.
```
