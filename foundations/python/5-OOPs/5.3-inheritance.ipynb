# Inheritance; it exists to avoid rewriting code when one class is a specialized version of another.

```python
class Animal:
    "..."
    def __init__(self, name):
        self.name = name

    def eat(self):
        "..."
        print(f"{self.name} is eating")

    def sleep(self):
        "..."
        print(f"{self.name} is sleeping")


class Dog(Animal):              # Dog inherits from Animal
    "Dog(Animal) means: Dog is a specialized Animal.It automatically gets everything Animal has, without rewriting eat()/sleep(), and adds its own new behavior (bark())."
    def bark(self):
        "..."
        print(f"{self.name} says Woof!")


d = Dog("Rex")
d.eat()      # inherited 
d.sleep()    # inherited 
d.bark()     # defined in Dog 

# Animal = parent class / base class / superclass
# Dog = child class / derived class / subclass
```

```python
class Rat(Animal):
    "..."
    pass  #do not keep class empty, if it doesn't have anything we either need to add methods specific to rat or just use pass statement(basically means nothing)
```

```python
# Overriding — redefining a parent's method in the child
class Cat(Animal):
    "..."
    def eat(self):
        print(f"{self.name} eats fish quietly")   # overrides Animal's eat()

c = Cat("Whiskers")
c.eat()   #Cat's version runs, not Animal's
```

```python
# super() — calling the parent's version from inside the child
class Dogie(Animal):
    "..."
    def __init__(self, name, breed):
        super().__init__(name)     # runs Animal's __init__ to set self.name
        self.breed = breed         # then adds Dog's own new attribute

d = Dogie("Rex", "Labrador")
print(d.name, d.breed)   # 

# Without super().__init__(name), you'd have to manually rewrite self.name = name again 
# super() lets you reuse the parent's setup instead of duplicating it.
```
