## Encapsulation; keep an object's internal data protected, and only allow controlled access through defined methods — rather than letting outside code directly mess with internal values.

```python
class BankAccount:
    "..."
    def __init__(self, balance):
        self.__balance = balance     # double underscore → "private"

    def deposit(self, amount):
        "..."
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        "..."
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        "..."
        return self.__balance


acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())    # 1300

print(acc.__balance)        # AttributeError — can't access directly
```

without encapsulation, anyone could do acc.balance = -99999 directly, corrupting your data with no validation. 
-> By forcing access through deposit()/withdraw(), you control how the data can change —> e.g., preventing negative withdrawals.

# Python's privacy levels (convention-based, not strictly enforced like other languages)
-> self.balance      # public — accessible anywhere
-> self._balance      # protected — convention: "internal use, avoid touching from outside"
-> self.__balance     # private — name gets mangled, hard (not impossible) to access from outside

#the underscores are a strong signal of intent, not an unbreakable wall.
