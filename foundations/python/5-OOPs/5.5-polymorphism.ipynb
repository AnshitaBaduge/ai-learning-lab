#  Polymorphism; "same method name, different behavior depending on the object." 
-> Lets you treat different objects uniformly, while each responds in its own appropriate way.

```python
class Animal:
    def __init__(self, name):
        self.name = name
        
class Dog(Animal):
    def make_sound(self):
        print("Woof")

class Cat(Animal):
    def make_sound(self):
        print("Meow")

class Cow(Animal):
    def make_sound(self):
        print("Moo")


animals = [Dog("Rex"), Cat("Whiskers"), Cow("Bella")]

for animal in animals:
    animal.make_sound()
```

-> Every object has a make_sound() method — same name — but each class defines its own version
-> Python automatically calls the correct one for each object type. You didn't need to: if type == Dog: ... elif type == Cat: ... — polymorphism handles that branching naturally.
