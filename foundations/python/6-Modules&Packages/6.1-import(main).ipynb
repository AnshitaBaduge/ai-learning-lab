## Importing modules in python: Modules and Packages

In python, modules and packages help organize and reuse code. 
A Module in python is a separate file with some reusable code, we use these modules to organize our code into diiferent files.

#imagine writing one giant .py file with 5000 lines of code — every function, every class, all in one place. It becomes impossible to find anything, impossible to reuse pieces elsewhere, and impossible for teams to work on it together without stepping on each other.
-> so we split code into separate .py files, each handling a related chunk of functionality. Each such file is called a module. 
-> A module is nothing exotic — any .py file is automatically a module the moment you can import it elsewhere.

```python
# main.py

import math_utils

print(math_utils.add(5, 3))         
print(math_utils.subtract(5, 3))    
print(math_utils.pi)                
```

Above:
#import math_utils — Python looks for a file called math_utils.py in the same directory, loads it, and makes everything inside it accessible

#math_utils.add(...) — access using dot notation, same pattern as calling a method — because the module itself acts like a container object holding all its functions/variables

```python
import math
math.sqrt(16)  # Import the whole module
```

```python
from math import sqrt,pi  # Import specific things directly (no need for module prefix)
print(sqrt(64))
print(pi)
```

```python
import numpy as np  # Import with a nickname (alias) — very common with popular libraries
np.sqrt(25)
np.mean([1,2,3,4])
```

```python
from math import *   # Import everything (generally discouraged — unclear where names come from)
sqrt(225)
```

#once you have many related modules, you group them into a folder, that folder is called a package.
Structure example:
my_project/
│
├── main.py
└── package/                  ← this folder is a package
    ├── __init__.py          ← marks the folder as a package(it's usually an empty file, but its presence is what tells Python "treat this folder as a package, not just a random folder." (In modern Python 3, this file is technically optional in many cases, but including it is still standard good practice, especially for clarity and compatibility.))
    ├── maths.py
    └── square.py

#__pycache__
-> Python automatically creates the first time you run or import a .py file. Inside, you'll find .pyc files (e.g., math_utils.cpython-312.pyc).

->Python is an interpreted language, but to speed things up, it secretly compiles your .py code into a lower-level format called bytecode the first time it runs, and saves that compiled version in __pycache__. Next time you import the same module (and it hasn't changed), Python reuses the cached bytecode instead of re-translating your source code from scratch — making imports faster

-> you never create or touch this folder yourself — Python manages it entirely automatically. It's safe to delete anytime (Python just regenerates it when needed), and it should never be manually edited or committed to Git (we'll cover .gitignore for exactly this reason when we reach Git).
