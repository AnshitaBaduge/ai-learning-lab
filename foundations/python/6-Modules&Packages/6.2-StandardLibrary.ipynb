## Standard Library Overview

Python's Standard Library is a vast collection of modules and packages that come bundled with python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard library.

```python
import array
arr = array.array('i',[1,2,3,4,5])
print(arr)
```

```python
import math
print(math.sqrt(16))     
print(math.pi)           
print(math.floor(4.7))   
print(math.ceil(4.2))
```

```python
import random
print(random.randint(1,20))   # random number between 1 and 10
print(random.choice(['A','B','C','D']))   # randomly picks one
```

```python
### File and Directory Access
import os
print(os.getcwd())  # Get current working directory
print(os.listdir())  # List all files and directories in current directory
#os.makedirs('test_direct')

## High level operations on files and collection of files
import shutil
shutil.copyfile('source.txt','destination.txt')
```

```python
## Data Serialization

import json
data = {'name':'Anshita','age':25}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv
import csv
with open('data.csv',mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Anshita',25])
    writer.writerow(['John',30])
    
with open('data.csv',mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime, timedelta   

now = datetime.now()
print("Current date and time:", now)

yesterday = now - timedelta(days=1)
print("Yesterday's date and time:", yesterday)

# Creating a timedelta object
delta = timedelta(days=5)
future_date = now + delta
print("Date after 5 days:", future_date)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## regular expression
import re
pattern = r'\d+'
text = 'There are 125 apples'
match = re.search(pattern,text)
print(match.group())  # here d+ pattern is geeting match, d is digits
```
