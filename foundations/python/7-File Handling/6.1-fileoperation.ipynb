File Operation - Read and Write Files

File Handling is a crucial part of any prgmming language. Python provides built-in funct and methods to read from and write to files, both text and binary.  

So far, everything you've done disappears the moment your program ends — variables live only in memory (RAM), which gets wiped when the script finishes. File handling lets your program save data permanently to disk, and read it back later, even after restarting.

```python
## Read a whole file
with open('example.txt','r') as file:
    content = file.read() # reads entire file as one string
    print(content)    
```

The open() function needs two things:

-> Filename — "notes.txt" (created automatically if it doesn't exist, in write mode)
-> Mode — what you intend to do with the file

Common modes:
-> "w" Write — creates file if missing, erases existing content if file exists
-> "a" Append — adds to the end, keeps existing content
-> "r" Read — file must already exist, or it errors
-> "x" Create — fails if file already exists

close(): while a file is open, Python holds a lock/connection to it — other programs (or parts of your own program) may not be able to access it properly, and unsaved data might not actually get written to disk yet. 

example: 
file = open("notes.txt", "r")
content = file.read()
print(content)
file.close()  # always close when done

```python
##Different ways to read:

with open("example.txt", "r") as file:
    print(file.readline())    # reads just one line

    lines = file.readlines()  # reads all lines into a list
    print(lines)              # ['line1\n', 'line2\n', ...]


with open('example.txt','r') as file:
    for line in file:         # loop directly over the file, line by line
        print(line.strip())   # strip() removes the newline character
        

# with the 'with' statement; file is automatically closed here, even if an error happened above
    
```

```python
## Writing a file (Overwriting)
with open('example.txt','w') as file:   # open in write mode
    file.write("Hello There!\n")        # files don't auto-add line breaks; you must include them yourself if you want separate lines.
    file.write("This is new line\n")
    file.write("*.*\n")
    
```

```python
## Writing a file ( Without Overwriting)
with open('example.txt','a') as file:        # Appending
    file.write("Append operation taking place!\n")
    file.write("Anshita Baduge this side.\n")
    file.write("^-^\n")
```

```python
## Writing a list of lines to a file.
lines = ['First line \n','Second line \n','Third line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## Binary Files

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
##Working with CSV files (common real-world format)

import csv

# Writing
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Marks"])
    writer.writerow(["Alex", 85])
    writer.writerow(["Priya", 92])

# Reading
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Working with JSON files (very common for structured data)

import json

data = {
    "name": "Alex", 
    "age": 25, 
    "is_learning_python": True}

# Writing a Python dict to a JSON file
with open("data.json", "w") as file:
    json.dump(data, file)               # json.dump(); Python object → JSON file

# Reading a JSON file back into a Python dict
with open("data.json", "r") as file:
    loaded_data = json.load(file)       # json.load(); JSON file → Python object
    print(loaded_data)          
    print(loaded_data["name"])  
```

json.dump() — Python object → JSON file. 
json.load() — JSON file → Python object. 
This format is everywhere: APIs, configs, data storage — worth being very comfortable with it.
