# Iterators
what makes for loops work on so many different things?

for x in [1, 2, 3]:        # works on a list
    print(x)

for ch in "hello":          # works on a string
    print(ch)

for k in {"a": 1, "b": 2}:  # works on a dict
    print(k)

completely different data types, yet the same 'for' loop syntax works on all of them. Internally, Python doesn't care what the data type is — it only asks: "can I get an iterator from this thing?"

-> Iterable — anything you can loop over (lists, strings, dicts, tuples, sets, files). Technically: an object that has an __iter__() method.
-> Iterator — the actual object that does the looping, remembering position, and produces one value at a time via __next__().
-> ex: 'for' grabs a "bookmark" (iterator) from the "book" (iterable) behind the scenes.

```python
numbers = [10, 20, 30]

it = iter(numbers)     # get the iterator (the "bookmark")

print(next(it))    
print(next(it))    
print(next(it))    
print(next(it))    # StopIteration error — no more items!
```

```python
## Building your own iterator
class CountUpTo:
    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self          # the object itself is the iterator

    def __next__(self):
        if self.current > self.limit:
            raise StopIteration     # signals "no more items"
        value = self.current
        self.current += 1
        return value


counter = CountUpTo(3)
for num in counter:
    print(num)
```
