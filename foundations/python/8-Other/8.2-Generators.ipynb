# Generators
writing a class with __iter__/__next__ every time is a lot of boilerplate for something simple. 
Generators solve this with far less code.

```python
def count_up_to(limit):
    current = 1
    while current <= limit:
        yield current       # yield is the key: it automatically makes this function behave like a full iterator, without you writing __iter__/__next__/StopIteration yourself.
        current += 1

for num in count_up_to(3):
    print(num)
```

```python
#yield vs return
def normal_function():
    print("Start")
    return 1
    print("This never runs")   # function ends completely at return

def generator_function():
    print("Start")
    yield 1
    print("Resumed after first yield")
    yield 2
    print("Resumed after second yield")
    yield 3
    
    
gen = generator_function()
print(next(gen))

print(next(gen))

print(next(gen))

# Each next() call resumes exactly where it paused, runs until the next yield, and pauses again. 
# return ends everything immediately; yield pauses and remembers state — this is the fundamental difference.
```

```python
## generators matter bcoz of memory efficiency reason (very important, real-world relevant)

# Builds the ENTIRE list in memory immediately — bad for huge n
def get_squares_list(n):
    return [x**2 for x in range(n)]

# Generates ONE value at a time, on demand — memory-efficient
def get_squares_gen(n):
    for x in range(n):
        yield x**2
        
# If n = 10,000,000, the list version tries to build and store 10 million numbers in memory right now. 
# The generator produces just one at a time, only when asked (next() or a for loop pulls it) — so memory usage stays tiny no matter how large n is. 
# This is essential for large datasets, reading huge files line-by-line, or infinite sequences.
```

```python
squares_list = [x**2 for x in range(5)]     # list — built immediately: [0,1,4,9,16]
squares_gen  = (x**2 for x in range(5))     # generator — nothing built yet

print(squares_gen)          # <generator object ...>
print(next(squares_gen))    # 0
print(next(squares_gen))    # 1

# Just swapping [] for () turns a list comprehension into a lazy generator expression.
```
