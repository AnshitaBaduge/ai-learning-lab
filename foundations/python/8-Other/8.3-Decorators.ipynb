# Decorators
you often want to add extra behavior around a function — like logging, timing, or checking permissions — without modifying the function's actual code. 

We can solve this with decorators

```python
# step 1 — remember functions are objects too
def greet():
    print("Hello!")

say_hi = greet   # assign the function itself (no parentheses = don't call it) to another name
say_hi()       

# Functions in Python can be passed around, stored in variables, and even passed into other functions — just like any other value (string, int, list). 
```

```python
# Step 2 — a function that takes another function as input
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()                              # call the original function
        print("After the function runs")
    return wrapper


def say_hello():
    print("Hello!")

decorated = my_decorator(say_hello)
decorated()

# my_decorator takes a function in, builds a new function (wrapper) that adds extra behavior before/after, and returns that new function.
```

```python
# Step 3 — the @ syntax (syntactic sugar for exactly the above)
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper


@my_decorator
def say_hello():
    print("Hello!")

say_hello()

# @my_decorator above def say_hello(): is exactly equivalent to writing say_hello = my_decorator(say_hello)
# Python just gives you cleaner syntax for this common pattern.
```

```python
## Example
# timing how long a function takes
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(f"Took {end - start:.4f} seconds")
    return wrapper


@timer
def slow_function():
    time.sleep(2)
    print("Done!")

slow_function()

# This is the real value of decorators: reusable "wrapping" logic (timing, logging, access-checks) 
# you can slap onto any function with one line, without touching that function's internals.
```
