# Context Managers
some resources (files, network connections, database connections) need to be properly closed/released after use — otherwise you leak memory or lock files. Forgetting to close things is a very common bug.

```python
# old, error prone way
file = open("data.txt", "w")
file.write("Hello")
file.close()          # if an error happens above this line, file never closes!

# If something crashes between open() and close(), the file stays open — a real problem in larger programs.
# The with statement — guarantees cleanup automatically, even if an error occurs

with open("data.txt", "w") as file:
    file.write("Hello")
    
# file is automatically closed here — guaranteed, no matter what happens inside

# with is Python's context manager syntax — it automatically handles setup (opening) and teardown (closing), even if an exception occurs inside the block.
```

```python
# Building your own context manager
class MyContextManager:
    def __enter__(self):
        print("Entering the block")
        return self             # what "as X" receives

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting the block, cleanup happens here")


with MyContextManager() as m:
    print("Inside the block")

# __enter__ — runs when entering the with block (setup)
# __exit__ — runs when leaving the block, even if an error occurred inside (guaranteed cleanup)
```

```python
# context managers — using generators + a decorator
from contextlib import contextmanager

@contextmanager
def my_context():
    print("Entering the block")
    yield
    print("Exiting the block")

with my_context():
    print("Inside the block")

# Code before yield = setup (__enter__) 
# Code after yield = cleanup (__exit__)
# This is a much shorter way to write a context manager than a full class.
```
