## venv (Virtual Environment)
different projects often need different versions of the same package. Project A needs requests==2.20, Project B needs requests==2.31. If you install packages globally (system-wide), you can only have one version at a time — installing one breaks the other project.

-> a virtual environment is an isolated, self-contained Python setup for a single project — its own separate space for installed packages, completely independent from your system's global Python and from other projects.

-> Creating one: python3 -m venv venv   # creates a folder named venv (you can name it anything) containing a private copy of the Python interpreter and an empty space to install packages into

-> Activating it (you must do this every time you start working):
venv\Scripts\activate

-> Once activated, your terminal prompt usually changes to show (venv) at the start — signaling "you are now inside the virtual environment." Any pip install now installs only into this project's isolated space, not system-wide.
ex: (venv) pip install requests

-> Deactivating: deactivate
